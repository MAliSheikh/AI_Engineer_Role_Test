```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from pathlib import Path

# Load environment variables from the parent directory
env_path = Path.cwd().parent / ".env"
load_dotenv(dotenv_path=env_path)

OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")

# Initialize the OpenRouter client
client = OpenAI(
    api_key=OPENROUTER_API_KEY,
    base_url="https://openrouter.ai/api/v1"
)

def get_ai_response(prompt_text):
    response = client.chat.completions.create(
        model="gpt-4o", # You can change this to any OpenRouter model you prefer
        messages=[
            {"role": "user", "content": prompt_text}
        ],
        max_tokens=1000,
        temperature=0.7
    )
    return response.choices[0].message.content

print("Setup complete! OpenRouter client initialized.")
```

### Prompt 1: Social Media Post
**Techniques Used:** Role assignment, context, and strict output constraints.

```python
prompt_1 = """You are a trendy social media manager for a disruptive streetwear footwear brand. 
Write an Instagram caption for the launch of our new eco-friendly sneaker line, the "AeroStep."
Target audience: Gen Z and young millennials who care about sustainability and urban fashion.
Tone: Edgy, authentic, and hype-building.
Constraints: Include 3 relevant hashtags, a clear Call to Action (CTA) pointing to the link in our bio,
and keep it strictly under 75 words."""


print("PROMPT 1")
output_1 = get_ai_response(prompt_1)
print("AI OUTPUT:")
print(output_1)
```

```python
print('Original Prompt: ')
print("Write a social media post for our new shoe brand.")
print()

explanation_1 = """
Explanation of Techniques:
1. This rewrite uses role assignment to instantly ground the AI in a specific persona and vocabulary set.
2. It provides vital context (platform, specific product name, target audience) so the AI understands exactly what it is selling and to whom.
3. Finally, it uses strict output constraints (under 75 words, exactly 3 hashtags, specific CTA) to ensure the generated text is immediately usable for a real social media workflow.
"""

comparison_1 = """
Before/After Comparison
1. Weak Prompt Output: 'Hey everyone! Check out our brand new shoe brand... Visit our website to buy your pair today! 👟🔥 #shoes
2. Rewritten Prompt Output: 'Sustainable drip just leveled up. 🌱👟 Meet the AeroStep: 100% recycled materials... Tap the link in our bio to grab your pair before they sell out! 💨👇 #AeroStep
"""



print("REWRITTEN PROMPT 1\n" + prompt_1)
print(explanation_1)
print(comparison_1 + "\n")
```

### Prompt 2: Ad Copy Creativity
**Techniques Used:** Evaluation criteria and few-shot prompting.
*Note: We are injecting a dummy generic ad copy into the prompt below to test it.*

```python
prompt_2_template = """You are an award-winning copywriter known for clever, punchy, and emotive advertising.
I will provide a piece of generic ad copy. Your task is to rewrite it to be more "creative." 
For this task, "creative" specifically means: focusing on emotional benefits rather than features, using active verbs, and incorporating a sensory hook.

Example Input: "Buy our mattress. It is very soft and helps you sleep better."
Example Output: "Drift into the deepest sleep of your life. Our cloud-spun mattresses turn tossing and turning into a thing of the past."

Now, apply this same framework to rewrite the following ad copy:
{ad_copy}"""

dummy_ad_copy = "Buy our new noise-cancelling headphones. They block out sound very well and have a long battery life."
prompt_2 = prompt_2_template.format(ad_copy=dummy_ad_copy)


print("PROMPT 2")
output_2 = get_ai_response(prompt_2)
print("AI OUTPUT:")
print(output_2)
```

```python

print('Original Prompt: ')
print("Make our ad copy more creative.")
print()

explanation_2 = """
Explanation of Techniques:
1. The original prompt fails because 'creative' is entirely subjective, leaving the AI guessing at the intent.
2. The rewrite solves this by explicitly defining the evaluation criteria for creativity (emotional benefits, active verbs, sensory hooks).
3. It also employs few-shot prompting by providing a clear before-and-after example, giving the model a concrete semantic pattern to match.
"""

comparison_2 = """
Before/After Comparison
1. Weak Prompt Output: 'Our headphones have great sound quality and long battery life. Buy now.'
2. Rewritten Prompt Output: 'Escape into pure sound. Let every beat surround you while the world fades away—experience silence like never before.'
"""


print("REWRITTEN PROMPT 2: \n" + prompt_2)
print(explanation_2)
print(comparison_2 + "\n")
```

### Prompt 3: Campaign Brief Summarization
**Techniques Used:** Strict output formatting (markdown template) and negative constraints.
*Note: We are injecting a dummy campaign brief into the prompt below to test it.*

```python
prompt_3_template = """Act as a Senior Account Executive. Please read the following campaign brief and extract the most critical information into a highly structured executive summary. 

Format your response using the EXACT markdown structure below:
- Core Objective: (One sentence summary)
- Target Audience: (Bullet points of key demographics)
- Key Deliverables: (List format)
- Timeline & Budget: (Brief summary)

Do not include any conversational filler, introductory, or concluding fluff. Keep the tone strictly professional.
        
Campaign Brief:
{campaign_brief}"""

dummy_brief = """We are launching a new summer beverage called 'Citrus Splash'. The goal is to drive brand awareness and trial among college students and young professionals aged 18-24. We need a 3-month social media campaign (TikTok and Instagram), a localized influencer partnership program in 5 major college towns, and experiential sampling at 3 summer music festivals. The total budget for this activation is $150,000 and it needs to run from June 1st to August 31st."""

prompt_3 = prompt_3_template.format(campaign_brief=dummy_brief)


print("PROMPT 3")
output_3 = get_ai_response(prompt_3)
print("AI OUTPUT:")
print(output_3)
```

```python

print('Original Prompt: ')
print("Summarize this campaign brief.")
print()

explanation_3 = """
Explanation of Techniques:
1. This rewrite utilizes strict output formatting by providing a markdown template, ensuring the AI only extracts the exact data points an agency team cares about.
2. It also leverages negative constraints ('Do not include any conversational filler...') to prevent the AI from generating unwanted pleasantries.
3. This ensures the output is structured, professional, and immediately usable in a real-world agency workflow.
"""

comparison_3 = """
Before/After Comparison
1. Weak Prompt Output: 'This campaign is about a summer drink targeting young people with social media marketing...'
2. Rewritten Prompt Output: '- **Core Objective:** Drive awareness and trial for Citrus Splash among young audiences...'
"""


print("REWRITTEN PROMPT 3\n" + prompt_3)
print(explanation_3)
print(comparison_3 + "\n")
```
